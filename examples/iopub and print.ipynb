# kernel directives

A kernel directive is a comment in the top line of code.

Lets define a function that we'll reuse in several contexts. Note - this example requires ipywidgets to be installed first.

```python
async def demo():
    import threading

    import anyio
    from ipywidgets import HTML, IntSlider

    print(threading.current_thread())
    slider = IntSlider()
    value = HTML()
    slider.observe(lambda c: print(f"slider: {c['new']} thread: {threading.current_thread().name}"), names="value")
    value.observe(lambda c: print(f"value: {c['new']} thread: {threading.current_thread().name}"), names="value")
    display(slider)
    display(value)
    while True:
        await anyio.sleep(0.1)
        val = str(slider.value)
        if value.value != val:
            value.value = val
            # print(value.value, end='\r')
        if slider.value == 100:
            return "Done"
```

Lets run this code in the cell first. Move the slider to the right so that it completes.

```python
await demo()
```
